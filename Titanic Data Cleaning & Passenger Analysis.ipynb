```python
plt.figure(figsize=(6,4))

sns.countplot(x="Survived", data=df)

plt.title("Survival Count")
plt.xlabel("Survived")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(x="Sex", data=df)

plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(x="Pclass", data=df)

plt.title("Passenger Class Distribution")
plt.xlabel("Passenger Class")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["Age"], bins=20, kde=True)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["Fare"], bins=20, kde=True)

plt.title("Fare Distribution")
plt.xlabel("Fare")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(x="Sex", hue="Survived", data=df)

plt.title("Survival by Gender")
plt.xlabel("Gender")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(x="Pclass", hue="Survived", data=df)

plt.title("Survival by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(10,6))

numeric_df = df.select_dtypes(include=["number"])

sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")

plt.show()
```

Summary

The Titanic dataset was analyzed using Python in Google Colab. The dataset was first cleaned by handling missing values, removing duplicate records, and treating outliers using the IQR method. The data was then filtered, sorted, and grouped to analyze passenger information based on age, gender, and passenger class.

Survival statistics showed that female passengers and first-class passengers had higher survival rates than male passengers and lower-class passengers. Various visualizations, including count plots, histograms, box plots, and a correlation heatmap, were created to understand the distribution of data and survival patterns.

Overall, this project demonstrated the complete process of Exploratory Data Analysis (EDA), including data cleaning, statistical analysis, grouping, filtering, and visualization, helping to derive meaningful insights from the Titanic dataset.
