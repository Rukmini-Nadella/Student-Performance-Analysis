```python
print("Total Sales:", df["Sales"].sum())
print("Average Sales:", df["Sales"].mean())
print("Maximum Sales:", df["Sales"].max())
print("Minimum Sales:", df["Sales"].min())

print("Total Profit:", df["Profit"].sum())
print("Average Profit:", df["Profit"].mean())
print("Maximum Profit:", df["Profit"].max())
print("Minimum Profit:", df["Profit"].min())
```

```python
region_sales = df.groupby("Region")[["Sales", "Profit"]].sum()
region_sales
```

```python
category_sales = df.groupby("Category")[["Sales", "Profit"]].sum()
category_sales
```

```python
category_sales = df.groupby("Category")["Sales"].sum()

plt.figure(figsize=(6,4))
category_sales.plot(kind="bar", color=["skyblue","orange","green"])
plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.show()
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])

monthly_sales = df.groupby(df["Order Date"].dt.to_period("M"))["Sales"].sum()

plt.figure(figsize=(12,5))
monthly_sales.plot()
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.hist(df["Sales"], bins=30, color="purple")
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.scatter(df["Sales"], df["Profit"], alpha=0.5)
plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.show()
```

Executive Summary

The Superstore sales dataset was analyzed to evaluate sales performance across different regions and product categories. Data cleaning was performed by checking for missing values and duplicates. Sales and profit statistics were calculated using Pandas and NumPy.

The analysis showed that the West region generated the highest sales and profit, making it the strongest-performing region. Among product categories, Technology achieved the highest sales and profit, while Furniture generated comparatively lower profit despite high sales.

Visualizations such as bar charts, line charts, histograms, and scatter plots helped identify sales trends, distribution patterns, and the relationship between sales and profit.

Recommendations
Focus marketing and inventory on the Technology category.
Continue investing in the West region due to its strong performance.
Review pricing and discount strategies for the Furniture category to improve profitability.
Use monthly sales trends to plan inventory and promotional campaigns.
